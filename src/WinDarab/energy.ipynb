```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load the data
df = pd.read_csv('20251206_matt_fwkilla_new.csv')
df.columns = df.columns.str.strip()

# print(df)
print(df['xtime [s]'])
```

```python
# Plotting
plt.figure(figsize=(12, 6))

time = df['xtime [s]'].values
power = df['WD_inverterPower[W]'].values
motor_rpm = df['PM_motorSpeed[rpm]'].values

wheel_radius = 0.2032  # example: 18" tire ≈ 0.2032 m radius
gear_ratio = 11/40

# Wheel RPM
wheel_rpm = motor_rpm * gear_ratio

# Convert RPM → rad/s
wheel_rad_s = wheel_rpm * 2*np.pi / 60

# Linear velocity (m/s)
vehicle_speed = wheel_rad_s * wheel_radius

distance = np.trapezoid(vehicle_speed, time)
print("Logged distance (m):", distance)

energy_logged = np.trapezoid(power, time)  # Joules

energy_per_meter = energy_logged / distance
energy_1km = energy_per_meter * 1000  # Joules
energy_1km_kWh = energy_1km / 3.6e6

print("Energy for 1 km (kWh):", energy_1km_kWh)
```

```python
energy_consumed = np.trapezoid(np.clip(power, 0, None), time)
energy_regen = np.trapezoid(np.clip(power, None, 0), time)

print("Consumed (kWh):", energy_consumed / 3.6e6)
print("Regen (kWh):", abs(energy_regen) / 3.6e6)
```
